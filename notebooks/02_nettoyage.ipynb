```python
import pandas as pd

df_google = pd.read_csv("../data/raw/entreprises_google_places.csv")
df_osm = pd.read_csv("../data/raw/entreprises_osm.csv")
df_brut = pd.concat([df_google, df_osm], ignore_index=True)

print(f"Total brut (avant nettoyage) : {len(df_brut)} lignes")
print("\nValeurs manquantes (brut) :")
print(df_brut.isna().sum())
```

```python
df_cleaned = pd.read_excel("../data/cleaned/entreprises_cleaned.xlsx")
print(f"Total après fusion + dédoublonnage + filtre catégories : {len(df_cleaned)} lignes")

print("\nRépartition par quartier (calculée par proximité GPS) :")
print(df_cleaned["quartier"].value_counts())
```
