```python
import pandas as pd

df = pd.read_csv("../data/processed/entreprises_final.csv")

print(f"Nombre total d'entreprises analysées : {len(df)}")
df[["note", "nombre_avis"]].describe()
```

```python
print("Répartition par domaine :")
df["categorie_recherchee"].value_counts()
```

```python
print("Note moyenne par quartier :")
df.groupby("quartier")["note"].mean().round(2).sort_values(ascending=False)
```

```python
correlation = df[["note", "nombre_avis"]].corr().iloc[0, 1]
print(f"Corrélation note / nombre d'avis : {correlation:.3f}")
print("Une corrélation faible/négative suggère qu'une petite entreprise")
print("avec peu d'avis peut avoir une très bonne note, et inversement.")
```
